```python
import os
import pyodbc
from dotenv import load_dotenv

# Asegurarse de cargar las variables de entorno
load_dotenv()

# Cargar credenciales desde variables de entorno con valores por defecto acordes a tu entorno
server = os.getenv('DB_SERVER', r'(localdb)\MSSQLLocalDB')
database = os.getenv('DB_NAME', 'DaviviendaSimulacion')
driver = os.getenv('DB_DRIVER', 'ODBC Driver 18 for SQL Server')

# Crear cadena de conexión agregando TrustServerCertificate=yes
conn_str = (
    f'DRIVER={{{driver}}};'
    f'SERVER={server};'
    f'DATABASE={database};'
    f'Trusted_Connection=yes;'
    f'TrustServerCertificate=yes;'
)

try:
    conn = pyodbc.connect(conn_str)
    print(f"Conexión exitosa a la base de datos '{database}' en '{server}'.")
    conn.close()
except Exception as e:
    print(f"Error al conectar a la base de datos: {e}")
```
